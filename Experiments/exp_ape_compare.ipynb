```python

# -----------------------------
# 3) Utilities
# -----------------------------
def get_single_and_combo_dicts(metric: str):
    if metric.upper() == "RMSE":
        # return S_RPEt_RMSE, C_RPEt_RMSE
        return S_RPEr_RMSE, C_RPEr_RMSE
        # return S_RMSE, C_RMSE
    elif metric.upper() == "MEAN":
        return S_MEAN, C_MEAN
    else:
        raise ValueError("Metric must be 'RMSE' or 'MEAN'")

def classify(seq, combo, singles, combos, treat_missing_single_as_win=True):
    if seq not in combos or combo not in combos[seq]:
        return -1
    a, b = combo.split('+')
    cv = combos[seq][combo]
    a_val = singles.get(seq, {}).get(a, None)
    b_val = singles.get(seq, {}).get(b, None)
    if a_val is None and b_val is None and not treat_missing_single_as_win:
        return -1
    beat_a = (cv < a_val) if a_val is not None else treat_missing_single_as_win
    beat_b = (cv < b_val) if b_val is not None else treat_missing_single_as_win
    if beat_a and beat_b:
        return 2
    if beat_a or beat_b:
        return 1
    return 0

def build_matrix(metric="RMSE", treat_missing_single_as_win=True):
    singles, combos = get_single_and_combo_dicts(metric)
    seqs = sorted(set(list(singles.keys()) + list(combos.keys())))
    all_combos = set()
    for s in combos:
        all_combos.update(combos[s].keys())
    combos_sorted = sorted(all_combos)
    mat = np.full((len(seqs), len(combos_sorted)), -1, dtype=int)
    for i, s in enumerate(seqs):
        for j, c in enumerate(combos_sorted):
            mat[i, j] = classify(s, c, singles, combos, treat_missing_single_as_win)
    return seqs, combos_sorted, mat

def save_color_grid(metric="RMSE", figsize=(4,3), treat_missing_single_as_win=True):
    seqs, combos_sorted, mat = build_matrix(metric, treat_missing_single_as_win)
    colors = ['#dedcdc', '#f4a582', '#a6dba0', '#1b7837']  # N/A, none, one, both
    cmap = ListedColormap(colors)
    bounds = [-1.5, -0.5, 0.5, 1.5, 2.5]
    norm = BoundaryNorm(bounds, cmap.N)

    fig = plt.figure(figsize=figsize)
    ax = plt.gca()
    im = ax.imshow(mat, cmap=cmap, norm=norm, aspect='auto')

    ax.set_xticks(np.arange(len(combos_sorted)))
    ax.set_xticklabels(combos_sorted, rotation=45, ha='right', fontsize=16)
    ax.set_yticks(np.arange(len(seqs)))
    ax.set_yticklabels(seqs, fontsize=16)
    # ax.set_xlabel("Feature Combinations", fontsize=12)
    # ax.set_ylabel("Sequences", fontsize=12)

    ax.set_xticks(np.arange(-.5, len(combos_sorted), 1), minor=True)
    ax.set_yticks(np.arange(-.5, len(seqs), 1), minor=True)
    ax.grid(which='minor', color='lightgray', linestyle='-', linewidth=0.5)
    ax.tick_params(which='minor', bottom=False, left=False)

    legend_patches = [
        mpatches.Patch(color='#1b7837', label='Multi beats both Single'),
        mpatches.Patch(color='#a6dba0', label='Multi beats a Single'),
        mpatches.Patch(color='#f4a582', label='Multi worse than both Single'),
        mpatches.Patch(color='#dedcdc', label='Excluded case'),
    ]
    # ax.legend(
    # handles=legend_patches,
    # loc='upper left',
    # bbox_to_anchor=(-0.01, 1.15),
    # ncol=len(legend_patches),
    # fontsize=16
    # )


    plt.tight_layout()
    base = f"./ResultFigs/combination_gain_grid_{metric}"
    png_path = base + ".png"
    pdf_path = base + ".pdf"
    plt.savefig(png_path, dpi=220, bbox_inches='tight')
    plt.savefig(pdf_path, bbox_inches='tight')
    plt.close(fig)
    return seqs, combos_sorted, mat, png_path, pdf_path

def matrix_to_dataframe(seqs, combos_sorted, mat):
    status_map = {2:'both', 1:'one', 0:'none', -1:'NA'}
    df = pd.DataFrame(mat, index=seqs, columns=combos_sorted).replace(status_map)
    return df

def save_latex_table(df, metric="RMSE"):
    def status_to_macro(s):
        m = str(s).strip().lower()
        if m == "both": return r"\cellboth"
        if m == "one":  return r"\cellone"
        if m == "none": return r"\cellnone"
        return r"\cellna"
    combos_sorted = list(df.columns)
    seqs = list(df.index)

    lines = []
    lines.append(r"% ==== Paste these into your preamble ====")
    lines.append(r"\usepackage{booktabs}")
    lines.append(r"\usepackage[table]{xcolor}")
    lines.append(r"\usepackage{graphicx}")
    lines.append(r"\usepackage{tikz}")
    lines.append(r"\newcommand{\cellboth}{\cellcolor[HTML]{1B7837}}")
    lines.append(r"\newcommand{\cellone}{\cellcolor[HTML]{A6DBA0}}")
    lines.append(r"\newcommand{\cellnone}{\cellcolor[HTML]{F4A582}}")
    lines.append(r"\newcommand{\cellna}{\cellcolor{white}}")
    lines.append(r"% ========================================")
    lines.append("")
    lines.append(r"\begin{table*}[t]")
    lines.append(r"\centering")
    lines.append(r"\scriptsize")
    lines.append(r"\setlength{\tabcolsep}{2pt}")
    lines.append(r"\renewcommand{\arraystretch}{1.2}")
    lines.append(r"\begin{minipage}{\textwidth}")
    lines.append(r"\centering")
    lines.append(r"\begin{tabular}{l" + "c"*len(combos_sorted) + "}")
    lines.append(r"\toprule")
    hdr = ["Seq ("+metric.upper()+")"] + [r"\rotatebox{60}{\strut "+c+r"}" for c in combos_sorted]
    lines.append(" & ".join(hdr) + r" \\")
    lines.append(r"\midrule")
    for s in seqs:
        row = [s] + [status_to_macro(df.loc[s, c]) for c in combos_sorted]
        lines.append(" & ".join(row) + r" \\")
    lines.append(r"\bottomrule")
    lines.append(r"\end{tabular}")
    lines.append(r"\vspace{0.3em}")
    lines.append(r"\begin{tikzpicture}[baseline]")
    lines.append(r"\definecolor{both}{HTML}{1B7837}")
    lines.append(r"\definecolor{one}{HTML}{A6DBA0}")
    lines.append(r"\definecolor{none}{HTML}{F4A582}")
    lines.append(r"\node[draw=black,fill=both,minimum width=10pt,minimum height=10pt] at (0,0) {}; \node[anchor=west] at (0.2,0) {Beats both (lower "+metric.upper()+r")};")
    lines.append(r"\node[draw=black,fill=one,minimum width=10pt,minimum height=10pt] at (3.9,0) {}; \node[anchor=west] at (4.1,0) {Beats $\geq$1};")
    lines.append(r"\node[draw=black,fill=none,minimum width=10pt,minimum height=10pt] at (6.8,0) {}; \node[anchor=west] at (7.0,0) {Beats none};")
    lines.append(r"\node[draw=black,fill=white,minimum width=10pt,minimum height=10pt] at (9.5,0) {}; \node[anchor=west] at (9.7,0) {N/A};")
    lines.append(r"\end{tikzpicture}")
    lines.append(r"\caption{Color grid using APE \textbf{"+metric.upper()+r"}: x-axis = two-feature combinations, y-axis = sequences. Deep green: combo lower than both singles; light green: lower than at least one; light red: lower than none; white: N/A. Missing single treated as win.}")
    lines.append(r"\label{tab:combo-gain-grid-"+metric.lower()+r"}")
    lines.append(r"\end{minipage}")
    lines.append(r"\end{table*}")
    tex_path = f"./ResultFigs/combination_gain_grid_{metric}_table.tex"
    with open(tex_path, "w", encoding="utf-8") as f:
        f.write("\n".join(lines))
    return tex_path

def save_status_csv(df, metric="RMSE"):
    csv_path = f"./ResultFigs/combination_gain_grid_{metric}.csv"
    df.to_csv(csv_path, encoding='utf-8')
    return csv_path

def summary_counts(df):
    counts = {}
    for seq in df.index:
        row = df.loc[seq]
        strict = (row == "both").sum()
        loose = ((row == "both") | (row == "one")).sum()
        total  = (row != "NA").sum()
        counts[seq] = {"beats_both": int(strict), "beats_ge1": int(loose), "total": int(total),
                       "strict_ratio": (strict/total if total>0 else np.nan),
                       "loose_ratio": (loose/total if total>0 else np.nan)}
    out = pd.DataFrame(counts).T
    return out

def global_summary(df):
    """Summarize across all sequences and combinations, including NA in total"""
    flat = df.values.flatten()
    total = len(flat)
    beats_both = np.sum(flat == "both")
    beats_ge1  = np.sum((flat == "both") | (flat == "one"))
    return {
        "beats_both": int(beats_both),
        "beats_ge1": int(beats_ge1),
        "total": int(total),
        "strict_ratio": beats_both / total if total > 0 else np.nan,
        "loose_ratio": beats_ge1 / total if total > 0 else np.nan,
    }
```
